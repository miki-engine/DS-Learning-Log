# Future Sales Forecasts (Basics of Linear Forecasting)

```python
# Sales trends over the past theree months
monthly_sales = [
    {"month": "January", "sales": 100000},
    {"month": "February", "sales": 110000},
    {"month": "March", "sales": 121000},
]

# calculate the growth rate
growth_rates = []
for i in range(1, len(monthly_sales)):
    prev = monthly_sales[i - 1]["sales"]
    curr = monthly_sales[i]["sales"]
    rate = (curr - prev) / prev
    growth_rates.append(rate)

avg_growth_rate = sum(growth_rates) / len(growth_rates)

def predict_next_sales(current_sales, avg_growth_rate):
    """Forecast next month's sales"""
    return current_sales * (1 + avg_growth_rate)

april_forecast = predict_next_sales(monthly_sales[-1]["sales"], avg_growth_rate)
print(f"April Sales Forecast: {april_forecast:,.0f} JPY")

TARGET = 150000
# How many months will it take to achieve the goal?
current = monthly_sales[-1]["sales"]
month = 0
while current < TARGET:
    current = predict_next_sales(current, avg_growth_rate)
    month += 1

print(f"Time to Goal: {TARGET:,} JPY ({month} Months Remaining)")

```
