# A/B Testing Analysis to Calculate CVR and Lift Metrics

```python
# A/B test session log（dummy data）
# Contains invalid data (conversion is greater than impression, negative values, etc)
ab_test_logs = [
    {"user_id": "U101", "group": "A", "impression": 100, "conversion": 5},
    {"user_id": "U102", "group": "B", "impression": 120, "conversion": 12},
    {"user_id": "U103", "group": "A", "impression": 80,  "conversion": 4},
    {"user_id": "U104", "group": "B", "impression": 150, "conversion": 18},
    {"user_id": "U105", "group": "A", "impression": 50,  "conversion": 6}, 
    {"user_id": "U106", "group": "B", "impression": -10, "conversion": 2},  # invalid data
    {"user_id": "U107", "group": "A", "impression": 90,  "conversion": 10},
    {"user_id": "U108", "group": "B", "impression": 110, "conversion": 11},
    {"user_id": "U109", "group": "A", "impression": 100, "conversion": 3},
    {"user_id": "U110", "group": "B", "impression": 130, "conversion": 14},
]

def is_valid(log):
    """
    Verify the log is valid
    Args:
        log (dict): AB testing log dictionary. Contains user_id, group, impression, conversion
    Returns:
        bool: True if impression >= conversion >= 0 and impression > 0
    """
    return log["impression"] >= log["conversion"] >= 0 and log["impression"] > 0
    
cleaned_logs = [log for log in ab_test_logs if is_valid(log)]

def calculate_cvr(group, logs):
    """
    Calculate the conversion rate for a specified group
    Args:
        group (str): Group name to calculate CVR for (e.g. "A" or "B")
        log (list): Data-cleaned AB testing logs. Each log is a dictionary
        Containing user_id, group, impression, conversion
    Returns:
        float: CVR of the specified group (conversion / impression)
    """
    group_logs = [log for log in logs if log["group"] == group]
    total_impression = sum(log["impression"] for log in group_logs)
    total_conversion = sum(log["conversion"] for log in group_logs)
    return total_conversion / total_impression


print(calculate_cvr("A", cleaned_logs))
print(calculate_cvr("B", cleaned_logs))
    

for log in ab_test_logs:
    print(log)
```
