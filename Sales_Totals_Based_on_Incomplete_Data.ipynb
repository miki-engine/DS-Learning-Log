# Sales Totals Based on Incomplete Data

```python
# Sales data by product（Partically missing）
products_data = [
    {"item": "Laptop", "sales": 150000},
    {"item": "Monitor", "sales": None},      # Missing data
    {"item": "Keyboard", "sales": 12000},
    {"item": "Mouse", "sales": 5000},
    {"item": "Webcam", "sales": None},      # Missing data
]

def is_valid_data(sales):
    """Check if the sales data is valid"""
    if sales == None:
        return False
    else:
        return True

total = 0    # Total valid data 
missing_data = []
for product in products_data:
    if is_valid_data(product["sales"]) == True:
        total += product["sales"]
    else:
        missing_data.append(product["item"])

    
print("=" * 50)
print("Total Sales")
print("=" * 50)
print(f"{total:,}円")
print(f"Skipped items: {missing_data}")
    
```
