# Identifying VIPs by Frequency & AOV

```python
# Customer Purchase History [name, amount]
purchase_history = [
    ["Tanaka", 15000],
    ["Sato", 8000],
    ["Tanaka", 12000],
    ["Suzuki", 5000],
    ["Sato", 20000],
    ["Takahashi", 50000],
    ["Tanaka", 10000],
    ["Suzuki", 4000],
]

customer_data = {}

for name, amount in purchase_history:
    if name not in customer_data:
        customer_data[name] = {"total": 0, "count": 0}
    customer_data[name]["total"] += amount
    customer_data[name]["count"] += 1

def calculate_average(total, count):
    """Calculate the average purchase amount per transaction"""
    return int(total / count)

for customer, data in customer_data.items():
    data["average"] = calculate_average(data["total"], data["count"])

# Determine if a user is a VIP
for customer, data in customer_data.items():
    if data["count"] >= 3 and data["average"] >= 10000:
        data["is_vip"] = True
    else:
        data["is_vip"] = False

print("=== 顧客一覧 ===")
for customer, data in customer_data.items():
    label = "[VIP]" if data["is_vip"] else "     "
    print(f"{customer:<10}{label} | 回数:{data['count']} | 平均:{data['average']:>6}円")

print("\n=== VIP顧客 ===")
for customer, data in customer_data.items():
    if data["is_vip"]:
        print(f"🎉{customer}🎉")

```
