# Identifying VIPs by Frequency & AOV

```python
# Customer Purchase History [name, amount]
purchase_history = [
    ["Tanaka", 15000],
    ["Sato", 8000],
    ["Tanaka", 12000],
    ["Suzuki", 5000],
    ["Sato", 20000],
    ["Takahashi", 50000],
    ["Tanaka", 10000],
    ["Suzuki", 4000],
]

customer_data = {}

for name, amount in purchase_history:
    if name not in customer_data:
        customer_data[name] = {"total": 0, "count": 0}
    customer_data[name]["total"] += amount
    customer_data[name]["count"] += 1

print(customer_data)
```
