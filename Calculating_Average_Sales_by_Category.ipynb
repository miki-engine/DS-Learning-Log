# Caltulating Average Sales by Category (Group Analysis)

```python
# Data by store (Area, Sales)
store_data = [
    {"area": "East", "sales": 150000},
    {"area": "West", "sales": 80000},
    {"area": "East", "sales": 200000},
    {"area": "North", "sales": 120000},
    {"area": "West", "sales": 95000},
    {"area": "North", "sales": 110000},
]

totals = {}    # {area: total sales}
counts = {}    # {area: number of stores}
for row in store_data:
    if row["area"] in totals:
        totals[row["area"]] += row["sales"]
        counts[row["area"]] += 1
    else:
        totals[row["area"]] = row["sales"]
        counts[row["area"]] = 1
    

def calculate_average(total, count):
    """Calculate the average sales"""
    return total / count

averages = {}    # {area: Average sales in the area}
for total in totals:
    averages[total] = calculate_average(totals[total], counts[total])


print(averages)
print(totals)
print(counts)
```
