# Calculation of Sales Target Achievement Rate

```python
# List of Sales Figures by Store
# branch: store name, sales: actual amount, target: target amount
branch_results = [
    {"branch": "Shibuya", "sales": 120000, "target": 100000},
    {"branch": "Shinjuku", "sales": 85000, "target": 90000},
    {"branch": "Ikebukuro", "sales": 150000, "target": 120000},
    {"branch": "Yokohama", "sales": 45000, "target": 60000},
]

def calculate_rate(sales, target):
    """Calculate rate from sales and target"""
    return sales / target * 100

print("=" * 50)
print("Sales Taget Achievement Rate")
print("=" * 50)
for result in branch_results:
    sales = result["sales"]
    target = result["target"]
    rate = calculate_rate(sales, target)
    
    if rate >= 100:
        judge = "PASS"
    else:
        judge = "RETRY"
    
    print(f"{result['branch']:<12} {rate:.1f}% [{judge}]")
```
