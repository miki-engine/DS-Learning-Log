# Analysis of Growth Rates by Category (Identifying Growth Stocks)

```python
# Sales data by category
sales_comparison = [
    {"category": "Gadgets", "last_month": 50000, "this_month": 65000},
    {"category": "Books", "last_month": 20000, "this_month": 18000},
    {"category": "Fashion", "last_month": 120000, "this_month": 132000},
    {"category": "Home", "last_month": 30000, "this_month": 45000},
]

def calculate_growth_rate(last, current):
    """Calculate the growth rate"""
    return (current - last) / last * 100

for sales in sales_comparison:
    sales["rate"] = calculate_growth_rate(sales["last_month"], sales["this_month"])

best_category = max(sales_comparison, key=lambda x: x["rate"])

print("=" * 50)
print("Growth Rates by Category")
print("=" * 50)

for sales in sales_comparison:
    print(f"{sales["category"]:<10}: {sales["rate"]:>5}%")

print("=" * 50)
print(f"The category with the highest Growth rate: {best_category["category"]}({best_category["rate"]}%)")
```
