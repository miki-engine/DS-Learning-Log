# Retention Analysis for Calculating Average Repeat Interval

```python
from datetime import datetime

# User activity log（Dummy data）
user_activity_logs = [
    {"user_id": "U001", "action": "signup", "date": "2026-01-01"},
    {"user_id": "U002", "action": "signup", "date": "2026-01-02"},
    {"user_id": "U001", "action": "purchase", "date": "2026-01-05"}, # U001's first purchase
    {"user_id": "U003", "action": "signup", "date": "2026-01-03"},
    {"user_id": "U002", "action": "purchase", "date": "2026-01-12"}, # U002's first purchase
    {"user_id": "U001", "action": "purchase", "date": "2026-01-15"}, # repeat purchase by U001
    {"user_id": "U003", "action": "purchase", "date": None},         # missing data
    {"user_id": "U004", "action": "signup", "date": "2026-01-05"},  # no repeat purchase
    {"user_id": "U002", "action": "purchase", "date": "2026-01-22"}, # repeat purchase by U002
]

def is_valid(log):
    """
    Verify whether the log is active
    Args:
        log: User activity log dictionary. Contains user_id, action, date.
    Returns:
        bool: Returns True if date is not None
    """
    return log["date"] is not None

user_activity_logs = [log for log in user_activity_logs if is_valid(log)]

# Convert date strings to datetime objects and sort by date
sorted_logs = sorted(user_activity_logs, key=lambda log: datetime.strptime(log["date"], "%Y-%m-%d"))

purchase_logs = []
for log in sorted_logs:
    if log["action"] == "purchase":
        purchase_logs.append(log)


print(purchase_logs)

for log in sorted_logs:
    print(log)
```
